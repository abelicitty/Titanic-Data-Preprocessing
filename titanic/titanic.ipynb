```python
import pandas as pd
```

```python
df_csv = pd.read_csv('titanic.csv')
titanic_df = df_csv
```

```python
df_xlsx = pd.read_excel('titanic.xlsx')
titanic_df = df_xlsx
```

```python
print("First 5 rows of the dataset:")
print(titanic_df.head())
```

```python
#PART 1 DATA CLEANING
```

```python
print("\nData structure and non-null counts:")
print(titanic_df.info())
```

```python
print("\nPercentage of Missing Values:")
missing_percent = titanic_df.isnull().sum() / len(titanic_df) * 100
print(missing_percent[missing_percent > 0].sort_values(ascending=False))
```

```python
titanic_df = titanic_df.drop('Cabin', axis=1)
print("Cabin column dropped.")
```

```python
median_age = titanic_df['Age'].median()
titanic_df['Age'] = titanic_df['Age'].fillna(median_age)
print(f"'Age' missing values filled with the median: {median_age}")
```

```python
mode_embarked = titanic_df['Embarked'].mode()[0] 
titanic_df['Embarked'] = titanic_df['Embarked'].fillna(mode_embarked)
print(f"'Embarked' missing values filled with the mode: {mode_embarked}")
```

```python
if titanic_df['Fare'].isnull().sum() > 0:
    median_fare = titanic_df['Fare'].median()
    titanic_df['Fare'] = titanic_df['Fare'].fillna(median_fare)
    print(f"'Fare' missing values filled with the median: {median_fare}")
else:
    print("'Fare' has no missing values.")
```

```python
print("\nFinal missing value check (should show 0 for Age, Embarked, and Fare):")
print(titanic_df.isnull().sum())
```

```python
#PART 2 DATA PREPROCESSING
```

```python
titanic_df['Title'] = titanic_df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)

rare_titles = ['Lady', 'Countess', 'Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona']
titanic_df['Title'] = titanic_df['Title'].replace(rare_titles, 'Rare')
titanic_df['Title'] = titanic_df['Title'].replace(['Mlle', 'Ms'], 'Miss')
titanic_df['Title'] = titanic_df['Title'].replace('Mme', 'Mrs')

print("\nDistribution of new 'Title' feature:")
print(titanic_df['Title'].value_counts())
```

```python
titanic_df['FamilySize'] = titanic_df['SibSp'] + titanic_df['Parch'] + 1

titanic_df['IsAlone'] = 0
titanic_df.loc[titanic_df['FamilySize'] == 1, 'IsAlone'] = 1
print("\nValue counts for 'IsAlone' feature:")
print(titanic_df['IsAlone'].value_counts())
```

```python
titanic_df = titanic_df.drop(['Name', 'Ticket', 'SibSp', 'Parch', 'PassengerId'], axis=1)
print("\nFinal set of columns after dropping:")
print(titanic_df.columns.tolist())
```

```python
titanic_df = pd.get_dummies(titanic_df, columns=['Sex', 'Embarked', 'Title', 'Pclass'], drop_first=True)

print("\nFinal Data Structure (Shape and Dtypes):")
print(titanic_df.shape)
print(titanic_df.dtypes)
print(titanic_df.head())
```
